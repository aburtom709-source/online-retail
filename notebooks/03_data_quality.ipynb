```python
# ============================================================
# DATA QUALITY - SILVER VALIDATION
# ============================================================

# IMPORT FUNCTIONS
from pyspark.sql.functions import col

# LOAD SILVER DATA
silver = spark.table("silver_retail")
```

```python
# DATA VALIDATION
total_records = silver.count()
```

```python
# NEGATIVE QUANTITIES
negative_quantity_count = (
    silver
    .filter(col("Quantity") < 0)
    .count()
)
```

```python
# NEGATIVE PRICE
negative_price_count = (
    silver
    .filter(col("UnitPrice") < 0)
    .count()
)
```

```python
# ZERO QUANTITY
zero_quantity_count = (
    silver
    .filter(col("Quantity") == 0)
    .count()
)
```

```python
# DATA NULL
null_customer_count = (
    silver
    .filter(col("CustomerID").isNull())
    .count()
)

null_description_count = (
    silver
    .filter(col("Description").isNull())
    .count()
)

return_count = (
    silver
    .filter(col("Quantity") < 0)
    .count()
)

sale_count = (
    silver
    .filter(col("Quantity") > 0)
    .count()
)

```

```python
# EXACT DUPLICATES
duplicate_count = (
    silver.count()
    - silver.dropDuplicates().count()
)
```

```python
# INVALID TOTAL
invalid_total_count = (
    silver
    .filter(
        col("Total") != col("Quantity") * col("UnitPrice")
    )
    .count()
)
```

```python
# DISPLAY RESULTS
print("DATA QUALITY REPORT")

print(f"Total records:              {total_records:,}")
print()

print("EXPECTED BUSINESS CONDITIONS")
print(f"Returns:                    {return_count:,}")
print(f"Sales:                      {sale_count:,}")
print(f"Null CustomerIDs:           {null_customer_count:,}")
print()

print("QUALITY ISSUES")
print(f"Negative prices:            {negative_price_count:,}")
print(f"Zero quantities:            {zero_quantity_count:,}")
print(f"Exact duplicates:           {duplicate_count:,}")
print(f"Invalid totals:             {invalid_total_count:,}")
print(f"Null descriptions:          {null_description_count:,}")
print()

```

```python
# VALIDATION
quality_checks_passed = (
    negative_price_count == 0
    and zero_quantity_count == 0
    and duplicate_count == 0
    and invalid_total_count == 0
    and null_description_count == 0
)
```

```python
# FINAL RESULT
if quality_checks_passed:
    print("✅ DATA QUALITY PASSED")
else:
    print("❌ DATA QUALITY FAILED")
```
