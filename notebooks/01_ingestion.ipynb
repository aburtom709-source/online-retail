```python
# ============================================================
# DATA INGESTION & RAW DATA (BRONZE)
# ============================================================

# INGESTION - BRONZE LAYER
source_path = "/Volumes/workspace/default/retail_source/"

# READ CSV FILES
df = (
    spark.read
    .format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load(source_path)
)
```

```python
# INGESTION VALIDATION
print(f"Record read from CSV: {df.count()}")

df.printSchema()

df.show(5)
```

```python
# SAVE AS DELTA - BRONZE
df.write.format("delta").mode("overwrite").saveAsTable("bronze_retail")

print("Bronze table created successfully")
```

```python
# FINAL VERIFICATION
bronze = spark.table("bronze_retail")

print(f"Bronze records count: {bronze.count()}")

bronze.show(5)
```
